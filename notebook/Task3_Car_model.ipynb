```python
# separates Features and Target
x=df.drop('Selling_Price',axis=1)
y=df['Selling_Price']

print('Features Shape:',x.shape)
print('Target Shape:',y.shape)
```

```python
x.head()
```

```python
# Removing unnecssary col
x=df.drop(columns=['Car_Name','Year','Selling_Price'])
print('Features Shape:',x.shape)
```

```python
x.head()
```

```python
# Apply one-hot encoding
x=pd.get_dummies(
    x,
    columns=['Fuel_Type',	'Selling_type',	'Transmission'],
    drop_first=True,
    dtype=int
)
```

```python
x.head()
```

```python
x.shape
```

```python
# Split the dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print('x_train shape:',x_train.shape)
print('x_test shape:',x_test.shape)
print('y_train shape:',y_train.shape)
print('y_test shape:',y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
num_cols=['Present_Price','Driven_kms','Owner','Car_Age']
x_train[num_cols]=scaler.fit_transform(x_train[num_cols])
x_test[num_cols]=scaler.transform(x_test[num_cols])
display(x_train.head())
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
# Comparison of Actual and Predicted Values
comparison=pd.DataFrame({
    'Actual Values':y_test.values,
    'Predicted Values':y_pred
})
comparison
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)
print('Model Evualtion Metrics: ')
print('Mean Absolute Error:',round(mae,2))
print('Mean Squared Error:',round(mse,2))
print('Root Mean Squared Error:',round(rmse,2))
print('r2_score:',round(r2,2))
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
rf.fit(x_train,y_train)
```

```python
y_pred=rf.predict(x_test)
comparison=pd.DataFrame({
    'Actual Values':y_test.values,
    'Predicted Values':y_pred
})
comparison
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
mae_rf=mean_absolute_error(y_test,y_pred)
mse_rf=mean_squared_error(y_test,y_pred)
rmse_rf=np.sqrt(mse_rf)
r2_rf=r2_score(y_test,y_pred)
print('Model Evualtion Metrics: ')
print('Mean Absolute Error:',round(mae_rf,2))
print('Mean Squared Error:',round(mse_rf,2))
print('Root Mean Squared Error:',round(rmse_rf,2))
print('r2_score:',round(r2_rf,2))
```

```python
models=['Linear Regression','Random Forest']
r2_scores=[r2,r2_rf]
plt.figure(figsize=(5,3))
sns.barplot(
    x=models,
    y=r2_scores
)
plt.title('r2_scores comparison')
plt.xlabel('Models')
plt.ylabel('r2_score')
plt.ylim(0,1.1)

for i,score in enumerate(r2_scores):
  plt.text(i,score+0.02,f'{score:.2f}',ha='center')
plt.show()
```

```python
# Extracting important features
feature_importance=pd.DataFrame({
    'Feature':x_train.columns,
    'Importance':rf.feature_importances_
})
feature_importance=feature_importance.sort_values(by='Importance',ascending=False)
display(feature_importance)
```

```python
plt.figure(figsize=(7,5))
sns.barplot(
    data=feature_importance,
    x='Importance',
    y='Feature'
)
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.title('Important Features and thier Importance')
plt.show()
```

```python
# Residuals Analysis
residuals=y_test-y_pred
plt.figure(figsize=(6,4))
sns.scatterplot(x=y_pred,y=residuals)
plt.axhline(y=0,linestyle='--',color='red')
plt.title('Residual Plot - Random Forest')
plt.xlabel('Predicted Selling Price (in lakhs)')
plt.ylabel('Residuals (in lakhs)')
plt.show()
```

```python

# Create model comparison table
model_comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest'],
    'MAE': [mae, mae_rf],
    'MSE': [mse, mse_rf],
    'RMSE': [rmse, rmse_rf],
    'R2 Score': [r2, r2_rf]
})

display(model_comparison.round(2))
```
